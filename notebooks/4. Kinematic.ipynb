```python
import os

# مسیر پوشه اصلی که حاوی 38 پوشه دیگر است
base_directory = r'C:\Mohsen Folder\Fatemeh Project\Analysis_Data 11\data\processed\Kinematic data (xsens suit)'

total_files = 0
num_folders = 0

# os.walk برای پیمایش درختی پوشه‌ها و فایل‌ها استفاده می‌شود
for dirpath, dirnames, filenames in os.walk(base_directory):
    # dirpath: مسیر پوشه فعلی
    # dirnames: لیستی از نام پوشه‌های زیرمجموعه در پوشه فعلی
    # filenames: لیستی از نام فایل‌های موجود در پوشه فعلی

    # تعداد پوشه‌هایی که پیدا کرده‌ایم را ثبت می‌کنیم (پوشه اصلی را حساب نمی‌کنیم)
    if dirpath != base_directory:
        num_folders += 1
        
    # تعداد فایل‌ها در پوشه فعلی را به شمارش کلی اضافه کنید
    total_files += len(filenames)

# نمایش نتایج
print(f"پوشه اصلی: {base_directory}")
print(f"تعداد کل پوشه‌های یافت شده (به جز پوشه اصلی): {num_folders}")
print(f"مجموع تعداد فایل‌ها در تمام پوشه‌ها: {total_files}")
```
