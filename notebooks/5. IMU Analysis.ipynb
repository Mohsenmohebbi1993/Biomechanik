> # IMU Analysis

An IMU (Inertial Measurement Unit) is a device composed of sensors such as an accelerometer, gyroscope, and often a magnetometer, used to measure the motion and orientation of an object in space. The accelerometer records linear acceleration, the gyroscope measures angular velocity, and the magnetometer provides magnetic field information to stabilize the yaw angle. By fusing these signals using algorithms like Kalman or Madgwick filters, the IMU can accurately estimate the object’s orientation and motion, making it essential in biomechanics, robotics, and gait analysis.

First read the **first 12 lines**, which contain the **sensor information**, and store them in a **DataFrame**.


```python
# Read 12 row in File

# Name EX: File
file_name = "MT_01200CEF_010-000_00B4D0BD.txt"

# Make Path: Make Path with Exampel File
full_path = os.path.join(input_path, file_name)

#---------------------------------------------
# Make one Empty datafram
first_12_row = pd.DataFrame({"Titel": [None]})

# Enter Data each line in Dataframe
with open(full_path, "r", encoding="utf-8") as f:
    for i in range(12):
        line = f.readline().rstrip("\n")
        first_12_row.loc[i, "Titel"] = line

# --- Clean first_12_row---
# Remove "// " in first of each Row
first_12_row["Titel"] = first_12_row["Titel"].str.replace("// ", "", regex=False)

# Split column to 2 columns by ":" 
first_12_row[["Titel", "Value"]] = first_12_row["Titel"].str.split(":", n=1, expand=True)

# Remove Space in first and end of each String
first_12_row["Titel"] = first_12_row["Titel"].str.strip()
first_12_row["Value"] = first_12_row["Value"].str.strip()

#---------------------------------------------
# Make one Dataframe from Data ( row 12 to end)
df_IMU = pd.read_csv(full_path,
                     skiprows=12,  # row 12 to end
                     sep=r"\t",   # separator= \t: one tab = one column
                     decimal=".")  # Float Number (EX: 14.25)
```

We have an Excel file named **map_sensor** with the **(.xlsx)** format that contains …

| Field     | Description                                      |
|-----------|--------------------------------------------------|
| ID        | Sensor location                                  |
| Location  | Physical location of the sensor on the body      |
| Name      | The name recognized by the OpenSim software      |

In this dataframe, the names that the OpenSim software uses to recognize the sensors are present.  
Now we merge these names into the **df_txt** dataframe.

With this approach, we get a complete DataFrame containing information from **every file**.  
Therefore, we can **filter by the person's name** to access the files that belong to that person, and then use a **for loop** to read all files for each person.

Additionally, we can determine:
- which **sensor** the file corresponds to,
- the **sensor position/body segment** (location),
- and the **file naming convention** used by the **OpenSim software**.


```python
# Path of Maping file
PATH_MAP_SENSOR = f"{input_path}\\map_sensor.xlsx"
# Read File that is maping file
MAP_SENSOR = pd.read_excel(PATH_MAP_SENSOR, sheet_name="Tabelle1")

# Merge df_txt and MAP_SENSOR and make tow columns "Location" and "Name"
# Name is name of opensim app
INFORMATION_TXT_FILE = pd.merge(df_txt,
                                MAP_SENSOR,
                                left_on="ID",
                                right_on="ID",
                                how="left")
```

By filtering based on `Name_User` and reading all files related to that person, we can build the **OpenSim DataFrame**.
From each file, we need the following columns:

- `time`  
- `sensor_name_q0`  
- `sensor_name_q1`  
- `sensor_name_q2`  
- `sensor_name_q3`  

These columns must appear **side by side in this exact format**.  
The **order of sensors** is not important, but the **order of data for each sensor** must remain consistent.


```python
INFORMATION_TXT_FILE[INFORMATION_TXT_FILE["Name_User"].eq("010-000")]
```

```python
# list of User
LIST_OF_USER = INFORMATION_TXT_FILE["Name_User"].unique()

# We create a **DataFrame** in which the required columns will be stored after reading each `.txt` file.
for i in LIST_OF_USER:
    df_all_data_txt = pd.DataFrame({"time":[]})

    MASK = INFORMATION_TXT_FILE[INFORMATION_TXT_FILE["Name_User"].eq(i)]
    
    for j in range(len(MASK)):
        name_of_sensor_for_opensim = MASK["Name"].iloc[j]

        path_of_file_txt = f"{input_path}\\{MASK["Main_Name"].iloc[j]}"
        print(path_of_file_txt)

        df_IMU = pd.read_csv(path_of_file_txt,
                             skiprows=12,  # row 12 to end
                             sep=r"\t",   # separator= \t: one tab = one column
                             decimal=".")
        
        df_all_data_txt[[f"{name_of_sensor_for_opensim}_q0",
                         f"{name_of_sensor_for_opensim}_q1",
                         f"{name_of_sensor_for_opensim}_q2",
                         f"{name_of_sensor_for_opensim}_q3"]] = df_IMU[["Quat_q0","Quat_q1", "Quat_q2","Quat_q3"]]
        
```
